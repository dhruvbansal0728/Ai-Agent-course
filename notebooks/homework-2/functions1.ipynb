```python

from agents import Agent, Runner, trace,function_tool
from openai.types.responses import ResponseTextDeltaEvent
```

```python
@function_tool
def add_numbers(a: int, b: int) -> int:
    """Add two numbers and return their sum."""
    return a + b
```

```python
from agents import Agent, Runner, trace

from addition_tool import add_numbers
from multiplication_tool import multiply_numbers
```

```python
math_agent = Agent(
    name="Math Assistant",
    instructions="You are a math assistant. Use the tools to perform calculations.",
    tools=[add_numbers, multiply_numbers]
)
```

```python
with trace("Math Assistant"):
    result = await Runner.run(
        math_agent,
        "Add 25 and 15, and multiply 10 and 5."
    )

print(result.final_output)
```
