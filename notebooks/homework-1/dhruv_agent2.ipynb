```python
import dotenv
import os

from openai import OpenAI

dotenv.load_dotenv()

if not os.environ.get("OPENAI_API_KEY"):
    print(
        """Error: OPENAI_API_KEY environment variable not set. Please copy the .env.template file as .env and fill it in.
    
    You can execute these commands in the terminal to get started:
    cp .env.template .env
    code .env
    """
    )

# Test OpenAI Access
print(
    OpenAI()
    .responses.create(
        model=os.environ["OPENAI_DEFAULT_MODEL"], input="Say: We are up and running!"
    )
    .output_text
)
```

```python
from agents import Agent, Runner, trace
from openai.types.responses import ResponseTextDeltaEvent
```

```python
travel_agent = Agent(
    name="Travel Planner",
    instructions="""
    You are a helpful travel planning assistant.

    - Suggest travel destinations.
    - Create day-wise travel itineraries.
    - Suggest famous attractions and activities.
    - Consider the user's budget and duration.
    - Recommend suitable transportation options.
    - Keep suggestions organized and easy to understand.
    - Do not invent current prices or opening hours.
    """
)
```

```python
with trace("Travel Planner Agent"):
    result = await Runner.run(
        travel_agent,
        "Plan a 3-day budget trip to Goa for two people."
    )

print(result.final_output)
```
