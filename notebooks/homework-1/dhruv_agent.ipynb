```python
import dotenv
import os

from openai import OpenAI

dotenv.load_dotenv()

if not os.environ.get("OPENAI_API_KEY"):
    print(
        """Error: OPENAI_API_KEY environment variable not set. Please copy the .env.template file as .env and fill it in.
    
    You can execute these commands in the terminal to get started:
    cp .env.template .env
    code .env
    """
    )

# Test OpenAI Access
print(
    OpenAI()
    .responses.create(
        model=os.environ["OPENAI_DEFAULT_MODEL"], input="Say: We are up and running!"
    )
    .output_text
)
```

```python
from agents import Agent, Runner, trace
from openai.types.responses import ResponseTextDeltaEvent
```

```python
interview_agent = Agent(
    name="Interview Preparation Assistant",
    instructions="""
    You are an interview preparation assistant.

    - Ask interview questions related to the user's subject.
    - Provide suitable answers with explanations.
    - Give tips for HR and technical interviews.
    - Explain difficult concepts in simple language.
    - Keep answers clear and concise.
    """
)
```

```python
with trace("Interview Preparation Agent"):
    result = await Runner.run(
        interview_agent,
        "Ask me 5 common Python interview questions with answers."
    )

print(result.final_output)
```
